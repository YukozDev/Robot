```python
#Imports

import sys
sys.path.append('/home/pi/project_demo/lib')

import time;
from Raspbot_Lib import Raspbot
from McLumk_Wheel_Sports import *

```

```python
#Initialisations

bot = Raspbot()
```

```python
def signal_stop() :
    #bot.Ctrl_BEEP_Switch(1)
    bot.Ctrl_WQ2812_ALL(1, 1)
    time.sleep(0.5)
    #bot.Ctrl_BEEP_Switch(0)
    bot.Ctrl_WQ2812_ALL(0, 0)
```

```python
def stop_and_clean() :
    stop()
    bot.Ctrl_BEEP_Switch(0)
    bot.Ctrl_WQ2812_ALL(0, 0)
```

```python
def read_ir_sensors():
    track_data = bot.read_data_array(0x0a, 1)
    track = int(track_data[0])
    return [
        (track >> 2) & 1,
        (track >> 3) & 1,
        (track >> 1) & 1,
        track & 1
    ]
```

```python
def control_bot_movement(ir_data):
    SPEED = 30
    l_ext = ir_data[0] == 0
    l_c = ir_data[1] == 0
    r_c = ir_data[2] == 0
    r_ext = ir_data[3] == 0
    print([l_ext, l_c, r_c, r_ext])
    if((l_c or r_c) and (not l_ext and not r_ext)):
        print("move_forward(SPEED)")
        move_forward(SPEED)
    elif(l_ext):
        print("rotate_left(SPEED)")
        rotate_left(SPEED/2)
    elif(r_ext):
        print("rotate_right(SPEED)")
        rotate_right(SPEED/2)
    elif(not l_ext and not l_c and not r_c and not r_ext):
        print("nothing")
        rotate_right(SPEED/3)
    else:
        print("stop()")
        stop()
```

```python
#Run
try :
    while True :
        ir_data = read_ir_sensors()
        control_bot_movement(ir_data)
        time.sleep(0.01)

except Exception as e:
    print(e)
    print("exception")
    stop_and_clean()


#del bot
```

```python
stop_and_clean()
del bot
```
