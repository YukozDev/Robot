```python
#Imports

import sys
sys.path.append('/home/pi/project_demo/lib')

import time;
from Raspbot_Lib import Raspbot
from McLumk_Wheel_Sports import *

```

```python
#Initialisations

bot = Raspbot()
```

```python
def advance(speed): 
    bot.Ctrl_Muto(0, speed)
    bot.Ctrl_Muto(1, speed)
    bot.Ctrl_Muto(2, speed)
    bot.Ctrl_Muto(3, speed)
```

```python
def reverse(speed): 
    bot.Ctrl_WQ2812_ALL(1,6)
    bot.Ctrl_Muto(0, -speed)
    bot.Ctrl_Muto(1, -speed)
    bot.Ctrl_Muto(2, -speed)
    bot.Ctrl_Muto(3, -speed)
```

```python
def stop():
    bot.Ctrl_Muto(0, 0)
    bot.Ctrl_Muto(1, 0)
    bot.Ctrl_Muto(2, 0)
    bot.Ctrl_Muto(3, 0)
    bot.Ctrl_WQ2812_ALL(1,0xFF0000)
    # bot.Ctrl_BEEP_Switch(1)
    # time.sleep(0.3)
    # bot.Ctrl_BEEP_Switch(0)

```

```python
def clear_lights():
    bot.Ctrl_WQ2812_ALL(0,0)

```

```python
def turn_left(speed):
    bot.Ctrl_Muto(0, speed)
    bot.Ctrl_Muto(1, speed)
    bot.Ctrl_Muto(2, -speed)
    bot.Ctrl_Muto(3, -speed)
    time.sleep(0.75)

def turn_right(speed):
    bot.Ctrl_Muto(0, -speed)
    bot.Ctrl_Muto(1, -speed)
    bot.Ctrl_Muto(2, speed)
    bot.Ctrl_Muto(3, speed)
    time.sleep(0.75)

```

```python
def find_line():
    track_data = bot.read_data_array(0x0a, 1)
    if track_data is None:
        return None

    track = int(track_data[0])
    return [
        (track >> 2) & 1,
        (track >> 3) & 1,
        (track >> 1) & 1,
        track & 1
    ]

```

```python
def kill():
    bot.Ctrl_BEEP_Switch(0)
    bot.Ctrl_Muto(0, 0)
    bot.Ctrl_Muto(1, 0)
    bot.Ctrl_Muto(2, 0)
    bot.Ctrl_Muto(3, 0)
    clear_light()
```

```python
try:
    while True:
        state = find_line()
        if state is None:
            time.sleep(0.05)
            continue

        far_left, left, right, far_right = state[0], state[1], state[2], state[3]

        if left == 0 and right == 0 and far_left == 0 and far_right == 0:
            stop()
        elif left == 1 and right == 1:
            advance(10)
        elif far_left == 1 and far_right == 1:
            advance(10)
        elif left == 1 and right == 0:
            turn_left(10)
        elif right == 1 and left == 0:
            turn_right(10)
        elif far_left == 1:
            turn_left(10)
        elif far_right == 1:
            turn_right(10)
        else:
            stop()

        time.sleep(0.05)

except KeyboardInterrupt:
    stop()
    clear_lights()
    print("Stopped by user.")
```
