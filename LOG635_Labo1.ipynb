```python
#Imports

import sys
sys.path.append('/home/pi/project_demo/lib')

import time;
from Raspbot_Lib import Raspbot
from McLumk_Wheel_Sports import *

```

```python
#Initialisations

MAX_EMPTY_ITERATIONS = 200
bot = Raspbot()
```

```python
def signal_stop() :
    #bot.Ctrl_BEEP_Switch(1)
    bot.Ctrl_WQ2812_ALL(1, 1)
    time.sleep(0.5)
    #bot.Ctrl_BEEP_Switch(0)
    bot.Ctrl_WQ2812_ALL(0, 0)
```

```python
def stop_and_clean() :
    stop()
    bot.Ctrl_BEEP_Switch(0)
    bot.Ctrl_WQ2812_ALL(0, 0)
```

```python
def read_ir_sensors():
    track_data = bot.read_data_array(0x0a, 1)
    track = int(track_data[0])
    return [
        (track >> 2) & 1,
        (track >> 3) & 1,
        (track >> 1) & 1,
        track & 1
    ]
```

```python
def control_bot_movement(ir_data, nextAction, counter):
    SPEED = 10
    l_ext = ir_data[0] == 0
    l_c = ir_data[1] == 0
    r_c = ir_data[2] == 0
    r_ext = ir_data[3] == 0
    #print([l_ext, l_c, r_c, r_ext])
    if((l_c or r_c) and (not l_ext and not r_ext)):
        #print("move_forward(SPEED)")
        move_forward(SPEED)
        nextAction = 0
        counter = 0
    elif(l_ext):
        #print("rotate_left(SPEED)")
        rotate_left(SPEED/3)
        nextAction = 1
        counter = 0
    elif(r_ext):
        #print("rotate_right(SPEED)")
        rotate_right(SPEED/3)
        nextAction = 2
        counter = 0
    elif(not l_ext and not l_c and not r_c and not r_ext):
        #if(counter > 0): 
        #    print([nextAction, counter])
        counter += 1
        if(counter >= MAX_EMPTY_ITERATIONS):
            stop()
            return nextAction, counter
        if(nextAction == 1):
            rotate_left(SPEED/3)
        elif(nextAction == 2):
            rotate_right(SPEED/3)
        else:
            #print("nothing")
            #nextAction == 3
            move_forward(SPEED/2)
    else:
        print("stop()")
        stop()
    return nextAction, counter
```

```python
#Run
try :
    nextAction = 0 
    counter = 0
    while counter <= MAX_EMPTY_ITERATIONS :
        ir_data = read_ir_sensors()
        nextAction, counter = control_bot_movement(ir_data, nextAction, counter)
        time.sleep(0.01)
    print([nextAction, counter])

except Exception as e:
    print(e)
    print("exception")
    stop_and_clean()

print("Allo")
stop_and_clean()
del bot
#del bot
```
