```python

import sys
sys.path.append('/home/pi/project_demo/lib')

import time;
from Raspbot_Lib import Raspbot
from McLumk_Wheel_Sports import *

def lire_ligne():
    """Retourne l'état des 4 capteurs de ligne : [extrême gauche, gauche, droite, extrême droite].
    Chaque valeur vaut 0 ou 1 (selon que le capteur voit une surface sombre ou claire)."""
    track = bot.read_data_array(0x0a, 1)[0]  # un octet, par exemple 0b0110 = 6
    #   bit 2 = extrême gauche, bit 3 = gauche, bit 0 = droite, bit 1 = extrême droite
    # (track >> n) & 1 : décale le bit n tout à droite, puis garde seulement ce bit
    return [(track >> 2) & 1,   # extrême gauche
            (track >> 3) & 1,   # gauche
            track & 1,          # droite
            (track >> 1) & 1]   # extrême droite
    
try:
    precedent = None
    while True:
        capteurs = lire_ligne()
        if capteurs != precedent:
            print(f"Capteurs ligne : {capteurs}", flush=True)
            precedent = lire_ligne()
            time.sleep(0.5)
except KeyboardInterrupt:  # clic sur ■ dans Jupyter
    print("Test arrêté")
```
